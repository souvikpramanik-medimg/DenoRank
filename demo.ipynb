# DenoRank demo
Gradient-free ranking of synthetic H&E images by diffusion denoising residual, compared against TracIn influence on one label fraction.

Requires a CUDA GPU.

```python
!pip install -q -r requirements.txt && pip install -q -e .
```

```python
import torch
from denorank.config import Config
from denorank.data.dataset import resolve_paths
from denorank.pipeline import run_one

cfg = Config(out_dir='runs', visualize=True)
paths = resolve_paths('data/CPM17')   # train/ validation/ test/ with images/ and masks/
device = torch.device('cuda')
```

## Run one setting (25% labels)

```python
rows = run_one(cfg, 'CPM17', paths, frac=0.25, seed=42, device=device)
```

```python
import pandas as pd
pd.DataFrame(rows)[['arm','n_syn','test_dice','test_iou','test_hd95']]
```
